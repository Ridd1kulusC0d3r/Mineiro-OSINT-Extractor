## Iniciar o servidor

```python
import subprocess, time, requests, os, signal

server = subprocess.Popen(
    ['npm', 'run', 'dev'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
)
time.sleep(5)

try:
    r = requests.get('http://127.0.0.1:3000', timeout=10)
    print('HTTP:', r.status_code)
    print('Mineiro está respondendo na porta 3000.')
except Exception as exc:
    print('Servidor ainda não respondeu:', exc)
```

## Abrir a interface pelo proxy do Colab

```python
from IPython.display import Javascript, display

display(Javascript("""
(async () => {
  const url = await google.colab.kernel.proxyPort(3000);
  console.log(url);
  window.open(url, '_blank');
})();
"""))
```

## Encerrar o servidor

Execute quando terminar a sessão.

```python
if 'server' in globals() and server.poll() is None:
    server.terminate()
    server.wait(timeout=10)
    print('Servidor encerrado.')
```
