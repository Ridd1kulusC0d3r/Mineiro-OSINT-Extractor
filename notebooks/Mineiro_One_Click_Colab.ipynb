## 1 · Preparar Mineiro
Execute esta célula uma vez por sessão do Colab. Ela clona ou atualiza o projeto, instala dependências e valida o Registry.

```python
import os, subprocess, pathlib
ROOT = pathlib.Path('/content/Mineiro-OSINT-Extractor')
if ROOT.exists():
    os.chdir(ROOT)
    subprocess.run(['git','checkout','main'], check=False)
    subprocess.run(['git','pull'], check=True)
else:
    subprocess.run(['git','clone','https://github.com/Ridd1kulusC0d3r/Mineiro-OSINT-Extractor.git', str(ROOT)], check=True)
    os.chdir(ROOT)
subprocess.run(['npm','install'], check=True)
subprocess.run(['npm','run','registry:validate'], check=True)
subprocess.run(['npm','run','registry:stats'], check=True)
print('\n[OK] Mineiro preparado em', ROOT)
```
